1) A company decided to give bonus of 5% to employee if his/her year of service is more than 5 years.

11) A shop will give discount of 10% if the cost of purchased quantity is more than 1000.
Ask user for quantity
Suppose, one unit will cost 100.
Judge and print total cost for user.

```python
quantity = int(input("Enter quantity: "))

cost = quantity * 100

if cost > 1000:
    discount = cost * 10 / 100
    total = cost - discount
else:
    total = cost

print("Total cost is:", total)
```

12) A school has following rules for grading system:

a. Below 25 - F

b. 25 to 45 - E

c. 45 to 50 - D

d. 50 to 60 - C

e. 60 to 80 - B

f. Above 80 - A

Ask user to enter marks and print the corresponding grade.

```python
marks = int(input("Enter your marks: "))

if marks < 25:
    print("Grade: F")
elif marks < 45:
    print("Grade: E")
elif marks < 50:
    print("Grade: D")
elif marks < 60:
    print("Grade: C")
elif marks <= 80:
    print("Grade: B")
else:
    print("Grade: A")
```

14)A student will not be allowed to sit in exam if his/her attendence is less than 75%.

Take following input from user

- Number of classes held

- Number of classes attended.

And print

- percentage of class attended

- Is student is allowed to sit in exam or not.


```python
held = int(input("Enter number of classes held: "))
attended = int(input("Enter number of classes attended: "))

percentage = (attended / held) * 100

print("\nPercentage of class attended:", round(percentage, 2), "%")

if percentage >= 75:
    print("Result: Student is allowed to sit in exam")
else:
    print("Result: Student is NOT allowed to sit in exam")
```

16) Write a program to check if a year is leap year or not.

If a year is divisible by 4 then it is leap year but if the year is century year like 2000, 1900, 2100 then it must be divisible by 400.

```python
year = int(input("Enter a year: "))

if (year % 400 == 0):
    print(year, "is a Leap Year")
elif (year % 100 == 0):
    print(year, "is NOT a Leap Year")
elif (year % 4 == 0):
    print(year, "is a Leap Year")
else:
    print(year, "is NOT a Leap Year")
```
